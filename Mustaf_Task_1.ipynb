#Importing the libraries

```python
import yfinance as yf
import pandas as pd
import numpy as np
```

#Defining the 10 stocks

```python
tickers = [
    "AAPL",   # Apple
    "MSFT",   # Microsoft
    "NVDA",   # NVIDIA
    "JPM",    # JPMorgan Chase
    "XOM",    # Exxon Mobil
    "AMZN",   # Amazon
    "META",   # Meta Platforms
    "GOOGL",  # Alphabet
    "TSLA",   # Tesla
    "UNH"     # UnitedHealth
]

print("Number of stocks:", len(tickers))
print(tickers)
```

#Choosing the historical period

```python
start_date = "2020-01-01"
end_date = "2026-07-26"
```

#Downloading the data from Yahoo Finance

```python
data = yf.download(
    tickers,
    start=start_date,
    end=end_date,
    auto_adjust=False,
    progress=False
)

print(data.head())
```

#Extracting adjusted closing prices

```python
prices = data["Adj Close"].copy()

print(prices.head())
```

#Inspecting the data

```python
print("Dataset shape:", prices.shape)

print("\nDate range:")
print("Start:", prices.index.min())
print("End:", prices.index.max())

print("\nMissing values:")
print(prices.isna().sum())

print("\nFirst 5 rows:")
print(prices.head())

print("\nLast 5 rows:")
print(prices.tail())
```

```python
prices.describe()
```

#Calculating daily log returns

```python
returns = np.log(prices / prices.shift(1))

returns = returns.dropna(how="all")

print(returns.head())
```

#Checking the returns

```python
print("Returns shape:", returns.shape)

print("\nMissing values:")
print(returns.isna().sum())

print("\nSummary statistics:")
print(returns.describe())
```

```python
# Save adjusted closing prices
prices.to_csv("Raw_Portfolio_Prices.csv")

# Save daily log returns
returns.to_csv("Portfolio_Returns.csv")

print("Files saved successfully!")
print("Raw_Portfolio_Prices.csv:", prices.shape)
print("Portfolio_Returns.csv:", returns.shape)
```

```python
# Create equal portfolio weights
weights = pd.Series(
    1 / len(tickers),
    index=tickers,
    name="Weight"
)

print(weights)
print("\nNumber of stocks:", len(weights))
print("Total weight:", weights.sum())
```

```python
weights.to_csv("Portfolio_Weights.csv", header=True)

print("Portfolio weights saved.")
```

```python
# Download S&P 500 market index data
market_data = yf.download(
    "^GSPC",
    start="2020-01-01",
    end="2026-07-26",
    auto_adjust=False,
    progress=False
)

print(market_data.head())
print("\nShape:", market_data.shape)
```

```python
# Extract S&P 500 adjusted closing prices
market_prices = market_data["Adj Close"].copy()

market_prices.columns = ["SP500"]

print(market_prices.head())
print("\nShape:", market_prices.shape)
print("\nMissing values:")
print(market_prices.isna().sum())
```

```python
market_prices.to_csv("Market_Index_Prices.csv")

print("Market index prices saved.")
```

```python
print("PHASE 1 SUMMARY")
print("-" * 40)

print("Stock price matrix:", prices.shape)
print("Stock return matrix:", returns.shape)
print("Number of portfolio weights:", len(weights))
print("Weights total:", weights.sum())
print("Market index matrix:", market_prices.shape)

print("\nStock price missing values:")
print(prices.isna().sum())

print("\nStock return missing values:")
print(returns.isna().sum())

print("\nMarket index missing values:")
print(market_prices.isna().sum())
```

```python
# Preserve Phase 1 data
clean_prices = prices.copy()
clean_market = market_prices.copy()

print("Stock prices:", clean_prices.shape)
print("Market index:", clean_market.shape)
```

```python
auto_adjust=False
```

Daily adjusted close prices obtained from Yahoo Finance are used. These prices account for corporate actions such as stock splits and dividends, avoiding the need for manual price adjustment.

```python
print("Stock dates:", len(clean_prices.index))
print("Market dates:", len(clean_market.index))

print("\nSame dates:",
      clean_prices.index.equals(clean_market.index))

print("\nStock start:", clean_prices.index.min())
print("Stock end:", clean_prices.index.max())

print("\nMarket start:", clean_market.index.min())
print("Market end:", clean_market.index.max())
```

```python
common_dates = clean_prices.index.intersection(clean_market.index)

clean_prices = clean_prices.loc[common_dates]
clean_market = clean_market.loc[common_dates]

print("Common trading dates:", len(common_dates))
print("Clean stock prices:", clean_prices.shape)
print("Clean market index:", clean_market.shape)

print("\nDates aligned:",
      clean_prices.index.equals(clean_market.index))
```

```python
# Phase 2: Data integrity checks

print("=== MISSING VALUES ===")
print("\nStock prices:")
print(clean_prices.isna().sum())

print("\nMarket index:")
print(clean_market.isna().sum())

print("\n=== DUPLICATE DATES ===")
print("Stock duplicate dates:", clean_prices.index.duplicated().sum())
print("Market duplicate dates:", clean_market.index.duplicated().sum())

print("\n=== INVALID PRICES ===")
print("Stock prices <= 0:", (clean_prices <= 0).sum().sum())
print("Market prices <= 0:", (clean_market <= 0).sum().sum())
```

```python
# Check consecutive unchanged prices
price_changes = clean_prices.diff()

stale_counts = (price_changes == 0).sum()

print("=== UNCHANGED PRICE OBSERVATIONS ===")
print(stale_counts.sort_values(ascending=False))
```

```python
# Display dates where a stock price was unchanged
for ticker in clean_prices.columns:
    stale_dates = clean_prices.index[clean_prices[ticker].diff() == 0]

    if len(stale_dates) > 0:
        print(f"\n{ticker}: {len(stale_dates)} unchanged observations")
        print(stale_dates.tolist())
```

```python
# Inspect unchanged-price observations with surrounding dates

for ticker in clean_prices.columns:
    stale_positions = np.where(clean_prices[ticker].diff().eq(0))[0]

    if len(stale_positions) > 0:
        print(f"\n{'='*50}")
        print(f"{ticker}")
        print(f"{'='*50}")

        for pos in stale_positions:
            start = max(0, pos - 1)
            end = min(len(clean_prices), pos + 2)

            print(clean_prices[ticker].iloc[start:end])
            print()
```

```python
clean_prices.loc[
    "2023-09-25":"2023-10-02",
    "AMZN"
]
```

```python
# Maximum consecutive unchanged-price runs

for ticker in clean_prices.columns:
    unchanged = clean_prices[ticker].diff().eq(0)

    groups = (~unchanged).cumsum()
    runs = unchanged.groupby(groups).sum()

    max_run = int(runs.max())

    print(f"{ticker}: maximum unchanged run = {max_run} day(s)")
```

Stale Price Check: Unchanged adjusted closing prices were identified and examined across all stocks. Most occurred for only one trading day, while AMZN had a maximum sequence of two consecutive unchanged observations. No prolonged stale-price patterns were identified; therefore, no observations were removed.

```python
# Calculate clean daily log returns
clean_returns = np.log(
    clean_prices / clean_prices.shift(1)
)

# Remove first row because no previous observation exists
clean_returns = clean_returns.dropna()

print("Clean returns shape:", clean_returns.shape)

print("\nMissing values:")
print(clean_returns.isna().sum())

print("\nInfinite values:")
print(np.isinf(clean_returns).sum())

print("\nFirst 5 observations:")
print(clean_returns.head())
```

```python
# Compare Phase 1 returns with Phase 2 clean returns

print("Same shape:",
      returns.shape == clean_returns.shape)

print("Same dates:",
      returns.index.equals(clean_returns.index))

print("Maximum difference:",
      (returns - clean_returns).abs().max().max())
```

```python
# Identify minimum and maximum daily log returns for each stock

extreme_summary = pd.DataFrame({
    "Min Return": clean_returns.min(),
    "Min Date": clean_returns.idxmin(),
    "Max Return": clean_returns.max(),
    "Max Date": clean_returns.idxmax()
})

print("=== EXTREME RETURN SUMMARY ===")
print(extreme_summary)
```

```python
# Identify returns with absolute movement greater than 10%

threshold = 0.10

for ticker in clean_returns.columns:
    extreme = clean_returns.loc[
        clean_returns[ticker].abs() > threshold,
        ticker
    ]

    print(f"\n{ticker}: {len(extreme)} observations > 10%")

    if len(extreme) > 0:
        print(extreme)
```

```python
# Show prices surrounding each stock's minimum and maximum return

for ticker in clean_returns.columns:

    min_date = clean_returns[ticker].idxmin()
    max_date = clean_returns[ticker].idxmax()

    print("\n" + "=" * 60)
    print(ticker)
    print("=" * 60)

    print("\nLargest negative return:")
    print("Date:", min_date)
    print("Return:", clean_returns.loc[min_date, ticker])

    min_pos = clean_prices.index.get_loc(min_date)
    print(
        clean_prices[ticker].iloc[
            max(0, min_pos - 1): min(len(clean_prices), min_pos + 2)
        ]
    )

    print("\nLargest positive return:")
    print("Date:", max_date)
    print("Return:", clean_returns.loc[max_date, ticker])

    max_pos = clean_prices.index.get_loc(max_date)
    print(
        clean_prices[ticker].iloc[
            max(0, max_pos - 1): min(len(clean_prices), max_pos + 2)
        ]
    )
```

Extreme Return Validation: Extreme daily log returns were identified and their corresponding adjusted prices and surrounding observations were examined. The extreme movements were consistent with genuine market or company-specific events, and no clear data-entry or extraction errors were detected. Therefore, no winsorization was applied, preserving genuine tail-risk information for subsequent risk modelling.

```python
# Align portfolio weights with return columns
aligned_weights = weights.reindex(clean_returns.columns)

print("=== ALIGNED PORTFOLIO WEIGHTS ===")
print(aligned_weights)

print("\nTotal weight:", aligned_weights.sum())

print("\nAny missing weights:",
      aligned_weights.isna().any())
```

```python
# Calculate weighted daily portfolio log returns
portfolio_returns = clean_returns.mul(
    aligned_weights,
    axis=1
).sum(axis=1)

portfolio_returns.name = "Portfolio_Return"

print("Portfolio observations:", len(portfolio_returns))
print("\nFirst 10 portfolio returns:")
print(portfolio_returns.head(10))

print("\nMissing values:", portfolio_returns.isna().sum())

print("\nPortfolio return statistics:")
print(portfolio_returns.describe())
```

```python
# Calculate 20-day rolling portfolio volatility
rolling_volatility = portfolio_returns.rolling(
    window=20
).std()

rolling_volatility.name = "Rolling_Volatility_20D"

print("Total observations:", len(rolling_volatility))
print("Missing observations:", rolling_volatility.isna().sum())

print("\nFirst 25 observations:")
print(rolling_volatility.head(25))

print("\nRolling volatility statistics:")
print(rolling_volatility.describe())
```

```python
rolling_volatility_annual = rolling_volatility * np.sqrt(252)

rolling_volatility_annual.name = "Rolling_Volatility_20D_Annualized"

print(rolling_volatility_annual.describe())
```

```python
# Save Phase 2 outputs

clean_prices.to_csv("Clean_Price_Matrix.csv")
clean_returns.to_csv("Clean_Return_Matrix.csv")

portfolio_returns.to_frame().to_csv(
    "Portfolio_Returns.csv"
)

rolling_volatility.to_frame().to_csv(
    "Rolling_Volatility_20D.csv"
)

rolling_volatility_annual.to_frame().to_csv(
    "Rolling_Volatility_20D_Annualized.csv"
)

print("Phase 2 files saved successfully.")
```

```python
print("=== PHASE 2 FINAL VALIDATION ===")

print("Clean prices:", clean_prices.shape)
print("Clean returns:", clean_returns.shape)
print("Portfolio returns:", portfolio_returns.shape)
print("Rolling volatility:", rolling_volatility.shape)

print("\nClean price missing values:",
      clean_prices.isna().sum().sum())

print("Clean return missing values:",
      clean_returns.isna().sum().sum())

print("Portfolio return missing values:",
      portfolio_returns.isna().sum())

print("Rolling volatility valid observations:",
      rolling_volatility.count())

print("Weights total:", aligned_weights.sum())

print("Dates aligned:",
      clean_prices.index.equals(clean_market.index))
```
